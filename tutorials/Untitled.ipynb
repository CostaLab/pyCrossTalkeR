```python
from pycrosstalker import tools as cttl
from pycrosstalker import plots as ctpl
import pandas as pd
import numpy as np
import anndata as an
import os
```

```python
ctr = pd.DataFrame()
ctr['source'] = ["A"]*3+["B"]*3+["C"]*3+["D"]*3
ctr['target'] = ["B"]*3+["B","C","D"]+["C"]*3+["D"]*3
ctr['type_gene_A'] = 'Ligand'
ctr['type_gene_B'] = 'Receptor'
ctr['gene_A'] = ["a1","a2","a3","b1","b2","b3","a1","a2","a3","b1","b2","b3"]
ctr['gene_B'] = ["b1","b2","b3","c1","d2","c3","b1","b2","b3","c1","d2","c3"]
ctr['MeanLR'] = np.random.sample(12)
ctr['LRScore'] = ctr['MeanLR']
```

```python
ctr
```

```python
exp = ctr.iloc[0:6,:]
exp['MeanLR'] = exp['MeanLR']*2
```

```python
aux = an.AnnData()
```

```python
aux.uns['pycrosstalker'] = {"path":{}}
aux.uns['pycrosstalker']['path']['ctr'] = ctr 
aux.uns['pycrosstalker']['path']['exp'] = exp

```

```python
output = "output/Myelofibrosis_example"
if not os.path.isdir(output):
    os.mkdir(output)
adata = cttl.analise_LR(aux,save=False)
```

```python
ctpl.plot.plot_cci(graph=adata.uns['pycrosstalker']['results']["graphs"]["exp"],
        colors=adata.uns['pycrosstalker']['results']["colors"],
        plt_name='Control',
        coords=adata.uns['pycrosstalker']['results']["coords"],
        emax= None,
        leg= False,
        low= 0,
        high= 0, 
        ignore_alpha= False,
        log= False,
        efactor= 6,
        vfactor= 12, 
        pg= adata.uns['pycrosstalker']['results']["rankings"]["exp"]["Pagerank"],
        scale_factor=2.0,
        node_size= 2.5,
        )
```
